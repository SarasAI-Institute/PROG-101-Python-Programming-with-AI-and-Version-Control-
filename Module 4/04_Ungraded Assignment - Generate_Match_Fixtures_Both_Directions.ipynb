## Additional Tests

These tests verify that the fixture logic works with different numbers of teams.

> You do not need to modify this cell.

````python
# Additional fixture tests

def expected_fixtures(teams):
    expected = []

    for team1 in teams:
        for team2 in teams:
            if team1 != team2:
                expected.append(team1 + " vs. " + team2)

    return expected


test_cases = [
    ['A', 'B'],
    ['A', 'B', 'C'],
    ['A', 'B', 'C', 'D'],
    ['India', 'Sri Lanka', 'Bangladesh', 'Pakistan']
]

for test_teams in test_cases:
    expected = expected_fixtures(test_teams)

    # Every team plays every other team in both directions.
    assert len(expected) == len(test_teams) * (len(test_teams) - 1)

    # No team plays itself.
    for fixture in expected:
        team1, team2 = fixture.split(" vs. ")
        assert team1 != team2

print("✅ All additional fixture tests passed!")
````
