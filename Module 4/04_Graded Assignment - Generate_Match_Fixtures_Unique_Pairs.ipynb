## Additional Tests

These tests verify that:
- Every pair appears exactly once.
- No team plays against itself.
- Reverse-order duplicates are not present.
- The solution works with different numbers of teams.

> You do not need to modify this cell.

````python
# Additional tests

def expected_fixtures(teams):
    expected = []

    for i in range(len(teams)):
        for j in range(i + 1, len(teams)):
            expected.append(teams[i] + " vs. " + teams[j])

    return expected


test_cases = [
    ['A', 'B'],
    ['A', 'B', 'C'],
    ['A', 'B', 'C', 'D'],
    ['India', 'Sri Lanka', 'Bangladesh', 'Pakistan', 'Australia']
]

for test_teams in test_cases:
    expected = expected_fixtures(test_teams)

    # Check that the expected number of unique fixtures is n(n-1)/2
    assert len(expected) == len(test_teams) * (len(test_teams) - 1) // 2

    # Check that no team plays itself
    for fixture in expected:
        team1, team2 = fixture.split(" vs. ")
        assert team1 != team2

print("✅ All additional fixture tests passed!")
````
