## Additional Tests

These tests verify that your logic works with:

- Different numbers of students
- Different score values
- Students with empty score lists
- Students who are not present in the dictionary

> You do not need to modify this cell.

````python
# Additional tests

test_cases = [
    (
        {'Alice': [100, 90, 80, 70, 60]},
        'Alice',
        "Alice's average score : 80.0"
    ),
    (
        {'Bob': []},
        'Bob',
        "No scores available for Bob"
    ),
    (
        {'Alice': [90, 90], 'Bob': [80, 80]},
        'Charlie',
        "Student Charlie not found in records"
    ),
    (
        {'David': [75, 85, 95]},
        'David',
        "David's average score : 85.0"
    )
]

for test_scores, test_student, expected in test_cases:
    try:
        scores = test_scores[test_student]

        if len(scores) == 0:
            actual = "No scores available for " + test_student
        else:
            average = sum(scores) / len(scores)
            actual = test_student + "'s average score : " + str(average)

    except KeyError:
        actual = "Student " + test_student + " not found in records"

    assert actual == expected

print("✅ All additional tests passed!")
````

A dictionary lookup can raise a `KeyError` when the requested student does not exist.
