## Additional Tests

The following tests demonstrate the expected swapping behavior with different values.

> You do not need to modify this cell.

````python
# Additional test cases

test_cases = [
    (10, 20),
    (-5, 15),
    (0, 100),
    (999, -100)
]

for original_x, original_y in test_cases:
    test_x, test_y = original_x, original_y
    test_x, test_y = test_y, test_x

    assert test_x == original_y
    assert test_y == original_x

print("✅ All additional tests passed!")
````

## Key Concept

Python allows two variables to exchange values directly using tuple unpacking:

```python
x, y = y, x
```

This does **not** require an additional temporary variable.
