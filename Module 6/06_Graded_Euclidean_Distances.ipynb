## Hint

The Euclidean distance between two points can be calculated as:

```text
sqrt((x2 - x1)² + (y2 - y1)²)
```

You can avoid `sqrt()` by using the power operator:

```python
((x2 - x1) ** 2 + (y2 - y1) ** 2) ** 0.5
```

A possible approach is to use `map()` with a lambda function:

```python
map(lambda point: ..., points)
```

You can then convert the result to a list and round each distance to two decimal places.
