### Developer Menu

Your program should present:

```text
Developer Menu:
1. Load Song Data
2. View Songs Database
3. Delete a Song
4. Modify a Song
5. Exit
```

### Requirements

#### 1. Load Song Data

- Prompt the developer for the file name.
- Read the song information from the text/CSV file.
- Store the information in a nested dictionary.
- The **key** should be the artist name.
- The **value** should be a list of dictionaries containing:
  - Title
  - Artist
  - Album
  - Genre
  - Duration
- Handle file-reading errors gracefully.

#### 2. View Songs Database

Display the songs in a neat table containing:

- Title
- Artist
- Genre

If no data has been loaded, inform the developer and return to the menu.

#### 3. Delete a Song

Prompt for:

- Artist
- Song title

If the song exists, delete it and display:

```text
Deleted "<title>" by "<artist>" from the database.
```

If it does not exist:

```text
Song not found.
```

#### 4. Modify a Song

Prompt for:

- Artist
- Song title

If found, display its current details and allow the developer to change:

- Album
- Genre
- Duration

Pressing **Enter** without entering a value should keep the existing value.

After modification:

```text
Modified "<title>" by "<artist>".
```

#### 5. Exit

Display:

```text
Exiting the application. Goodbye!
```

Then exit the menu loop using `break`.

> **Important:** Do not use `exit()` to terminate the application. Use `break`.

> **Important:** Match the required prompts and messages as closely as possible because the original assignment uses automated checks.
