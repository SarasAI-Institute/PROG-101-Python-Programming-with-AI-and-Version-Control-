## Additional Tests

These tests cover:

````python
# Additional tests

test_cases = [
    (
        "Contact john123@example.com and jane_doe@test.org.",
        ['john123@example.com', 'jane_doe@test.org']
    ),
    (
        "There are no email addresses in this sentence.",
        []
    ),
    (
        "Email admin_01@company123.co.uk for support.",
        ['admin_01@company123.co.uk']
    ),
    (
        "Try test@example.com or user_99@domain.net today.",
        ['test@example.com', 'user_99@domain.net']
    )
]

pattern = r'(?<![A-Za-z0-9_])[A-Za-z0-9_]+@[A-Za-z0-9_]+(?:\.[A-Za-z]+)+(?![A-Za-z])'

for test_text, expected in test_cases:
    actual = re.findall(pattern, test_text)
    assert actual == expected

print("✅ All additional tests passed!")
````

## Hint

The `re` module provides functions for working with regular expressions.

A useful function is:

```python
re.findall(pattern, text)
```

It returns all matches found in the text as a list.

A simple pattern can be built from these parts:

```text
username @ domain . extension
```

For example:

```python
r'[A-Za-z0-9_]+@[A-Za-z0-9_]+\.[A-Za-z]+'
```

The `+` means "one or more" characters.

You may need to make the pattern more precise so that invalid email-like strings are not included.
